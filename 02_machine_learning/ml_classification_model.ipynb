```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
import warnings
warnings.filterwarnings('ignore')

# 1. Load the Data (Aapke file ka naam yahan hai)
df = pd.read_csv('sample_output.csv') 

# 2. Create the Training Data (Examples for the model)
training_data = [
    ("NVIDIA announces new Blackwell GPU architecture", "Hardware"),
    ("How AI is Helping Care Teams Get Patients the Right Care", "AI Software"),
    ("NVIDIA expands open source commitment with OpenShell", "AI Software"),
    ("NVIDIA announces new planned investments in Korean startups", "Business"),
    ("NVIDIA partners with tech giants for new data center", "Hardware"),
    ("Vision AI and robotics pick what's worth recovering", "AI Software"),
    ("NVIDIA financial results exceed expectations", "Business"),
    ("New GPU drivers improve gaming performance", "Hardware")
]
train_texts = [text for text, label in training_data]
train_labels = [label for text, label in training_data]


# 3. Convert Text to Numbers (TF-IDF)
vectorizer = TfidfVectorizer(stop_words='english')
X_train = vectorizer.fit_transform(train_texts)

# 4. Train the Model
model = MultinomialNB()
model.fit(X_train, train_labels)
print("Model trained successfully!")

# 5. Make Predictions
df['combined_text'] = df['Title'].astype(str) + " " + df['Description'].astype(str)
X_test = vectorizer.transform(df['combined_text'])
df['Predicted Category'] = model.predict(X_test)

# 6. Save the final file for Power BI
df_clean = df.drop(columns=['combined_text'])
df_clean.to_csv('nvidia_news_ml_output.csv', index=False)

print("\nPredictions completed! Here is the preview:")
print(df_clean[['Title', 'Predicted Category']].head())
```
